# Task 3: Data Cleaning with Python

## Objective
To transform a messy Titanic dataset into a clean, analysis-ready dataset using Python, Pandas, and NumPy.

## Dataset
Titanic passenger training dataset (`train.csv`).

## Cleaning Steps
1. Data quality assessment
2. Missing value handling
3. Duplicate detection and removal
4. Data standardisation
5. Outlier detection
6. Data type correction
7. Before-and-after quality comparison
8. Export the cleaned dataset

```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

df = pd.read_csv("train.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python

# Display basic dataset information
print("DATASET SHAPE")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nDATA TYPES AND NON-NULL COUNTS")
df.info()

print("\nFIRST 5 ROWS")
display(df.head())
```

```python

# Data Quality Report - Before Cleaning

quality_report = pd.DataFrame({
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isnull().sum(),
    "Missing %": (df.isnull().mean() * 100).round(2),
    "Unique Values": df.nunique(),
})

print("Missing values by column:")
display(quality_report)

print("Duplicate rows:", df.duplicated().sum())
print("Total rows:", len(df))

print("\nNumeric column summary:")
display(df.describe().T)
```

```python

# Check numeric ranges for possible anomalies

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    print(f"\n{col}")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
```

## Initial Data Quality Observations

- The dataset contains passenger information from the Titanic training dataset.
- Missing values must be checked and handled column by column.
- Duplicate rows will be identified and removed if present.
- Categorical values will be checked for inconsistent formatting.
- Numeric ranges will be reviewed to identify possible anomalies and outliers.
- Data types will be verified before exporting the cleaned dataset.

```python
print("Missing values per column:")
display(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
display(df.dtypes)
```

## Missing Data Handling Decisions

1. **Age:** Missing values will be replaced with the median because age is numeric and the median is robust to outliers.
2. **Cabin:** Missing values will be labelled as "Unknown". Since most cabin values are missing, deleting these rows would cause substantial data loss.
3. **Embarked:** Missing values will be replaced with the mode because this is a categorical column and only two values are missing.
4. **Other columns:** No imputation is required because they contain no missing values.
5. **Duplicates:** No exact duplicate rows were found, so no rows will be removed during duplicate cleaning.

```python

# Keep a copy of the original dataset for comparison
df_before = df.copy()

# Handle missing values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Cabin"] = df["Cabin"].fillna("Unknown")
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Remove exact duplicate rows
duplicates_removed = df.duplicated().sum()
df = df.drop_duplicates().copy()

print("Missing values after treatment:")
display(df.isnull().sum())

print("\nDuplicate rows removed:", duplicates_removed)
print("Rows remaining:", len(df))
```

## Data Standardisation

Categorical values will be checked for inconsistent capitalization and extra spaces. The `Sex` and `Embarked` columns will be standardized to consistent formatting. The `Cabin` column will retain "Unknown" for missing cabin information. Passenger IDs will be stored as strings because they identify records rather than represent quantities.

```python

# Standardize text columns
text_columns = ["Name", "Sex", "Ticket", "Cabin", "Embarked"]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

# Standardize categorical values
df["Sex"] = df["Sex"].str.lower().replace({
    "male": "Male",
    "female": "Female"
})

df["Embarked"] = df["Embarked"].str.upper()

# Correct data types
df["PassengerId"] = df["PassengerId"].astype("string")
df["Survived"] = df["Survived"].astype("int64")
df["Pclass"] = df["Pclass"].astype("int64")
df["Age"] = pd.to_numeric(df["Age"], errors="coerce").astype("float64")
df["Fare"] = pd.to_numeric(df["Fare"], errors="coerce").astype("float64")
df["SibSp"] = df["SibSp"].astype("int64")
df["Parch"] = df["Parch"].astype("int64")

print("Updated data types:")
display(df.dtypes)

print("\nSex values:", df["Sex"].unique())
print("Embarked values:", df["Embarked"].unique())
```

```python

# Detect outliers using the IQR method
outlier_results = []

for col in ["Age", "Fare", "SibSp", "Parch"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    count = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    outlier_results.append({
        "Column": col,
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Outlier Count": int(count)
    })

outlier_report = pd.DataFrame(outlier_results)
display(outlier_report)
```

## Outlier Detection and Treatment

The Interquartile Range (IQR) method was used to identify potential outliers in Age, Fare, SibSp, and Parch.

- **Age:** Outliers were retained because unusually young or old passenger ages may be valid.
- **Fare:** Outliers were retained because high ticket prices may reflect passenger class or ticket arrangements.
- **SibSp:** Outliers were retained because passengers could travel with larger groups of siblings or spouses.
- **Parch:** Outliers were retained because some passengers travelled with larger groups of parents or children.

No outliers were removed or capped because the detected values were not confirmed data-entry errors. Retaining them preserves potentially meaningful information.

```python

# Before vs. after data quality summary

summary = pd.DataFrame({
    "Metric": [
        "Row Count",
        "Column Count",
        "Missing Cells",
        "Duplicate Rows",
        "Data Types"
    ],
    "Before Cleaning": [
        len(df_before),
        df_before.shape[1],
        int(df_before.isnull().sum().sum()),
        int(df_before.duplicated().sum()),
        df_before.dtypes.astype(str).to_dict()
    ],
    "After Cleaning": [
        len(df),
        df.shape[1],
        int(df.isnull().sum().sum()),
        int(df.duplicated().sum()),
        df.dtypes.astype(str).to_dict()
    ]
})

display(summary[["Metric", "Before Cleaning", "After Cleaning"]])
```

```python

# Final validation
print("Final dataset shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nFinal data types:")
display(df.dtypes)

# Save the cleaned dataset
output_file = "Titanic_Cleaned.csv"
df.to_csv(output_file, index=False)

print(f"\nCleaned dataset saved as: {output_file}")
```
